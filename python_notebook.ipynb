```python
import pandas as pd
import numpy as np

# Load cleaned dataset
file_name = "Cleaned_Sample_Dataset.xlsx"
df = pd.read_excel(file_name)

# Basic data check
df.columns = df.columns.str.strip()
df = df.drop_duplicates()

# Convert date column
df["event_date"] = pd.to_datetime(df["event_date"], errors="coerce")

# Convert numeric columns
numeric_columns = [
    "fit_score",
    "search_time_ms",
    "search_result_rank",
    "job_viewed",
    "candidate_viewed"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# ---------------------------------------------------------
# 1. Overall Company Funnel
# ---------------------------------------------------------

total_events = len(df)
unique_companies = df["company_id"].nunique()
unique_jobs = df["job_id"].nunique()
unique_students = df["student_id"].nunique()

job_views = df["job_viewed"].sum()
candidate_views = df["candidate_viewed"].sum()

job_view_rate = (job_views / total_events) * 100 if total_events > 0 else 0
candidate_view_rate = (candidate_views / total_events) * 100 if total_events > 0 else 0

company_funnel = pd.DataFrame({
    "Metric": [
        "Total Events",
        "Unique Companies",
        "Unique Jobs",
        "Unique Students",
        "Job Viewed",
        "Candidate Viewed",
        "Job View Rate (%)",
        "Candidate View Rate (%)"
    ],
    "Value": [
        total_events,
        unique_companies,
        unique_jobs,
        unique_students,
        job_views,
        candidate_views,
        round(job_view_rate, 2),
        round(candidate_view_rate, 2)
    ]
})

# ---------------------------------------------------------
# 2. Search Performance Analysis
# ---------------------------------------------------------

search_performance = df.groupby("search_source").agg(
    Total_Searches=("event_id", "count"),
    Job_Views=("job_viewed", "sum"),
    Candidate_Views=("candidate_viewed", "sum"),
    Average_Fit_Score=("fit_score", "mean"),
    Average_Search_Time_ms=("search_time_ms", "mean")
).reset_index()

search_performance["Job_View_Rate_%"] = (
    search_performance["Job_Views"] /
    search_performance["Total_Searches"] * 100
).round(2)

search_performance["Candidate_View_Rate_%"] = (
    search_performance["Candidate_Views"] /
    search_performance["Total_Searches"] * 100
).round(2)

search_performance["Average_Fit_Score"] = (
    search_performance["Average_Fit_Score"].round(2)
)

search_performance["Average_Search_Time_ms"] = (
    search_performance["Average_Search_Time_ms"].round(2)
)

# ---------------------------------------------------------
# 3. Job Category Analysis
# ---------------------------------------------------------

category_analysis = df.groupby("job_category").agg(
    Total_Events=("event_id", "count"),
    Unique_Jobs=("job_id", "nunique"),
    Job_Views=("job_viewed", "sum"),
    Candidate_Views=("candidate_viewed", "sum"),
    Average_Fit_Score=("fit_score", "mean"),
    Average_Search_Time_ms=("search_time_ms", "mean")
).reset_index()

category_analysis["Job_View_Rate_%"] = (
    category_analysis["Job_Views"] /
    category_analysis["Total_Events"] * 100
).round(2)

category_analysis["Candidate_View_Rate_%"] = (
    category_analysis["Candidate_Views"] /
    category_analysis["Total_Events"] * 100
).round(2)

category_analysis["Average_Fit_Score"] = (
    category_analysis["Average_Fit_Score"].round(2)
)

category_analysis["Average_Search_Time_ms"] = (
    category_analysis["Average_Search_Time_ms"].round(2)
)

# ---------------------------------------------------------
# 4. Region Analysis
# ---------------------------------------------------------

region_analysis = df.groupby("region").agg(
    Total_Events=("event_id", "count"),
    Unique_Companies=("company_id", "nunique"),
    Unique_Jobs=("job_id", "nunique"),
    Job_Views=("job_viewed", "sum"),
    Candidate_Views=("candidate_viewed", "sum"),
    Average_Fit_Score=("fit_score", "mean"),
    Average_Search_Time_ms=("search_time_ms", "mean")
).reset_index()

region_analysis["Job_View_Rate_%"] = (
    region_analysis["Job_Views"] /
    region_analysis["Total_Events"] * 100
).round(2)

region_analysis["Candidate_View_Rate_%"] = (
    region_analysis["Candidate_Views"] /
    region_analysis["Total_Events"] * 100
).round(2)

region_analysis["Average_Fit_Score"] = (
    region_analysis["Average_Fit_Score"].round(2)
)

region_analysis["Average_Search_Time_ms"] = (
    region_analysis["Average_Search_Time_ms"].round(2)
)

# ---------------------------------------------------------
# 5. Experience Level Analysis
# ---------------------------------------------------------

experience_analysis = df.groupby("experience_level").agg(
    Total_Events=("event_id", "count"),
    Job_Views=("job_viewed", "sum"),
    Candidate_Views=("candidate_viewed", "sum"),
    Average_Fit_Score=("fit_score", "mean"),
    Average_Search_Time_ms=("search_time_ms", "mean")
).reset_index()

experience_analysis["Job_View_Rate_%"] = (
    experience_analysis["Job_Views"] /
    experience_analysis["Total_Events"] * 100
).round(2)

experience_analysis["Candidate_View_Rate_%"] = (
    experience_analysis["Candidate_Views"] /
    experience_analysis["Total_Events"] * 100
).round(2)

experience_analysis["Average_Fit_Score"] = (
    experience_analysis["Average_Fit_Score"].round(2)
)

experience_analysis["Average_Search_Time_ms"] = (
    experience_analysis["Average_Search_Time_ms"].round(2)
)

# ---------------------------------------------------------
# 6. Company Type Analysis
# ---------------------------------------------------------

company_type_analysis = df.groupby("company_type").agg(
    Total_Events=("event_id", "count"),
    Unique_Companies=("company_id", "nunique"),
    Unique_Jobs=("job_id", "nunique"),
    Job_Views=("job_viewed", "sum"),
    Candidate_Views=("candidate_viewed", "sum"),
    Average_Fit_Score=("fit_score", "mean"),
    Average_Search_Time_ms=("search_time_ms", "mean")
).reset_index()

company_type_analysis["Job_View_Rate_%"] = (
    company_type_analysis["Job_Views"] /
    company_type_analysis["Total_Events"] * 100
).round(2)

company_type_analysis["Candidate_View_Rate_%"] = (
    company_type_analysis["Candidate_Views"] /
    company_type_analysis["Total_Events"] * 100
).round(2)

company_type_analysis["Average_Fit_Score"] = (
    company_type_analysis["Average_Fit_Score"].round(2)
)

company_type_analysis["Average_Search_Time_ms"] = (
    company_type_analysis["Average_Search_Time_ms"].round(2)
)

# ---------------------------------------------------------
# 7. Fit Score Analysis
# ---------------------------------------------------------

df["fit_category"] = pd.cut(
    df["fit_score"],
    bins=[0, 59, 79, 100],
    labels=["Low Fit", "Medium Fit", "High Fit"]
)

fit_analysis = df.groupby(
    "fit_category",
    observed=False
).agg(
    Total_Events=("event_id", "count"),
    Job_Views=("job_viewed", "sum"),
    Candidate_Views=("candidate_viewed", "sum"),
    Average_Search_Time_ms=("search_time_ms", "mean")
).reset_index()

fit_analysis["Job_View_Rate_%"] = (
    fit_analysis["Job_Views"] /
    fit_analysis["Total_Events"] * 100
).round(2)

fit_analysis["Candidate_View_Rate_%"] = (
    fit_analysis["Candidate_Views"] /
    fit_analysis["Total_Events"] * 100
).round(2)

fit_analysis["Average_Search_Time_ms"] = (
    fit_analysis["Average_Search_Time_ms"].round(2)
)

# ---------------------------------------------------------
# 8. Search Speed Analysis
# ---------------------------------------------------------

search_speed = df.groupby("search_source").agg(
    Minimum_Search_Time_ms=("search_time_ms", "min"),
    Average_Search_Time_ms=("search_time_ms", "mean"),
    Maximum_Search_Time_ms=("search_time_ms", "max")
).reset_index()

search_speed["Average_Search_Time_ms"] = (
    search_speed["Average_Search_Time_ms"].round(2)
)

# ---------------------------------------------------------
# 9. Data Quality Check
# ---------------------------------------------------------

quality_check = pd.DataFrame({
    "Check": [
        "Total Rows",
        "Total Columns",
        "Duplicate Rows",
        "Missing Values",
        "Minimum Fit Score",
        "Maximum Fit Score",
        "Minimum Search Time (ms)",
        "Maximum Search Time (ms)"
    ],
    "Value": [
        len(df),
        len(df.columns),
        df.duplicated().sum(),
        df.isnull().sum().sum(),
        round(df["fit_score"].min(), 2),
        round(df["fit_score"].max(), 2),
        df["search_time_ms"].min(),
        df["search_time_ms"].max()
    ]
})

# ---------------------------------------------------------
# 10. Freshness Check
# ---------------------------------------------------------

latest_event_date = df["event_date"].max()
earliest_event_date = df["event_date"].min()

freshness_check = pd.DataFrame({
    "Metric": [
        "Earliest Event Date",
        "Latest Event Date",
        "Total Event Days",
        "Data Availability"
    ],
    "Value": [
        earliest_event_date,
        latest_event_date,
        (latest_event_date - earliest_event_date).days + 1,
        "Available" if not df.empty else "No Data"
    ]
})

# ---------------------------------------------------------
# 11. Top Performing Jobs
# ---------------------------------------------------------

top_jobs = df.groupby("job_id").agg(
    Total_Searches=("event_id", "count"),
    Job_Views=("job_viewed", "sum"),
    Candidate_Views=("candidate_viewed", "sum"),
    Average_Fit_Score=("fit_score", "mean"),
    Average_Search_Time_ms=("search_time_ms", "mean")
).reset_index()

top_jobs["Candidate_View_Rate_%"] = (
    top_jobs["Candidate_Views"] /
    top_jobs["Total_Searches"] * 100
).round(2)

top_jobs["Average_Fit_Score"] = (
    top_jobs["Average_Fit_Score"].round(2)
)

top_jobs["Average_Search_Time_ms"] = (
    top_jobs["Average_Search_Time_ms"].round(2)
)

top_jobs = top_jobs.sort_values(
    by=["Candidate_Views", "Average_Fit_Score"],
    ascending=[False, False]
).head(10)

# ---------------------------------------------------------
# Save all analysis outputs into ONE Excel file
# ---------------------------------------------------------

output_file = "Python Analysis Result.xlsx"

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    company_funnel.to_excel(
        writer,
        sheet_name="Company Funnel",
        index=False
    )

    search_performance.to_excel(
        writer,
        sheet_name="Search Performance",
        index=False
    )

    category_analysis.to_excel(
        writer,
        sheet_name="Job Category",
        index=False
    )

    region_analysis.to_excel(
        writer,
        sheet_name="Region Analysis",
        index=False
    )

    experience_analysis.to_excel(
        writer,
        sheet_name="Experience Analysis",
        index=False
    )

    company_type_analysis.to_excel(
        writer,
        sheet_name="Company Type",
        index=False
    )

    fit_analysis.to_excel(
        writer,
        sheet_name="Fit Analysis",
        index=False
    )

    search_speed.to_excel(
        writer,
        sheet_name="Search Speed",
        index=False
    )

    quality_check.to_excel(
        writer,
        sheet_name="Data Quality",
        index=False
    )

    freshness_check.to_excel(
        writer,
        sheet_name="Freshness Check",
        index=False
    )

    top_jobs.to_excel(
        writer,
        sheet_name="Top Jobs",
        index=False
    )

print("Python analysis completed successfully.")
print("All analysis outputs are saved in one file:")
print(output_file)
```
